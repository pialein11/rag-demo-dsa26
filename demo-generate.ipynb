```python
from ollama import chat 

MODEL = "llama3.1:latest"

messages = [
   {"role": "user", "content": "Schreibe ein Herbstgedicht mit 5 Absätzen."}
]
```

```python
response = chat(model=MODEL, messages=messages)
```

```python
print(response.message.content) 
```

#### Context Size bei Llama 3.1 (Ollama)

- Llama 3.1 kann bis zu **128k Tokens** Kontext, Ollama nutzt standardmäßig aber viel weniger (z.B. 2048/4096, je nach Version/VRAM).
- Ist der Text länger, schneidet Ollama ihn **still ab**, ohne Fehlermeldung oder Warnung.
- **Folge:** Teile vom Prompt oder von den RAG-Dokumenten fehlen, das Modell antwortet trotzdem (evtl. falsch oder unvollständig).
- **Problem:** Man sieht nicht, *ob* und *was* abgeschnitten wurde → Kontextgröße mit `num_ctx` bewusst setzen.

### Kontext weiter schreiben 

```python
#Response des Systems in den Messages 
messages.append({"role": "assistant", "content": response.message.content}
                )
```

```python
# ev. nächste User-Message ergänzen 
messages.append({"role": "user", "content": "Kürze das Gedicht auf 3 Absätze."})
```

```python
# Neuen erweiterten Kontext an das Modell übergeben und weiter generieren lassen 
chat(
    model=MODEL,
    messages=messages
)
```
